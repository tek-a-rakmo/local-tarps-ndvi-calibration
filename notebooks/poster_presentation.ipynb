```python
import pandas as pd

# Step 1: Load the Excel file
# Ensure this file path is correct and the file exists.
file_path = r"D:\Capstone\EDA\poster_data.xlsx"
try:
    df_poster = pd.read_excel(file_path)
except FileNotFoundError:
    print(f"Error: The file was not found at {file_path}")
    # Exit or handle error appropriately if file not found
    exit()

# Step 2: Remove filename columns
df_poster = df_poster.drop(columns=['filename_pixel_2', 'filename_red', 'filename_white'], errors='ignore')

# Step 3: Keep only one date column (choosing 'date_pixel_2')
# Ensure 'date_pixel_2' is the correct column to keep and others exist to be dropped.
df_poster = df_poster.drop(columns=['date_red', 'date_white'], errors='ignore')

# Step 4: Remove band 1 and band 2 columns
columns_to_drop_bands = [
    'pixel_2_band_1', 'pixel_2_band_2',
    'red_tarp_band1', 'red_tarp_band2',
    'white_tarp_band1', 'white_tarp_band2'
]
df_poster = df_poster.drop(columns=columns_to_drop_bands, errors='ignore')

# Step 5: Rename the band columns
rename_map = {
    'pixel_2_band_3': 'pixel_2_red',
    'pixel_2_band_4': 'pixel_2_nir',
    'red_tarp_band3': 'red_tarp_red',
    'red_tarp_band4': 'red_tarp_nir',
    'white_tarp_band3': 'white_tarp_red',
    'white_tarp_band4': 'white_tarp_nir'
}
df_poster = df_poster.rename(columns=rename_map)

# Step 6: Display the transformed DataFrame (optional initial check)
print("DataFrame after initial cleaning (Steps 1-5):")
print(df_poster.head())
print("-" * 50)

# Step 7: Inspect the first few values of the 'date_pixel_2' column
# This step helps understand the format of the date column before cleaning.
print("Original 'date_pixel_2' column (first 5 values):")
if 'date_pixel_2' in df_poster.columns:
    print(df_poster['date_pixel_2'].head())
else:
    print("'date_pixel_2' column not found. Please check column names after Step 3.")
    exit() # Exit if the crucial date column is missing
print("-" * 50)

# Step 8: Clean the 'date_pixel_2' column.
# Assuming the format is like '020230515' (a prefix followed by 'YYYYMMDD'),
# we extract the last 8 characters.
# The original code had a syntax error here (missing ']').
print("Cleaning 'date_pixel_2' column...")
df_poster['date_pixel_2_cleaned'] = df_poster['date_pixel_2'].astype(str).str.strip()
# Handle potential cases like 'nan' or other non-numeric prefixes before slicing
# This ensures that we are trying to slice from a string that is expected to contain date info
df_poster['date_pixel_2_cleaned'] = df_poster['date_pixel_2_cleaned'].str[-8:]
print("Cleaned 'date_pixel_2_cleaned' (last 8 chars, first 5 values):")
print(df_poster['date_pixel_2_cleaned'].head())
print("-" * 50)

# Step 9: Convert the cleaned 'date_pixel_2_cleaned' into a proper datetime format
# The format '%Y%m%d' assumes the cleaned string is like '20230515'.
# 'errors=coerce' will turn unparseable dates into NaT (Not a Time).
print("Converting 'date_pixel_2_cleaned' to datetime objects...")
df_poster['datetime_pixel_2'] = pd.to_datetime(df_poster['date_pixel_2_cleaned'], format='%Y%m%d', errors='coerce')
print("Converted 'datetime_pixel_2' (first 5 values):")
print(df_poster['datetime_pixel_2'].head())
# Check for NaT values, which indicate parsing problems
if df_poster['datetime_pixel_2'].isnull().any():
    print("\nWarning: Some dates could not be parsed. Review 'date_pixel_2_cleaned' and the original 'date_pixel_2'.")
    print("Rows with NaT dates:")
    print(df_poster[df_poster['datetime_pixel_2'].isnull()][['date_pixel_2', 'date_pixel_2_cleaned']].head())
print("-" * 50)

# Step 10: Create a new column 'date' from 'datetime_pixel_2' (extract date part only)
print("Extracting date part into 'date' column...")
df_poster['date'] = df_poster['datetime_pixel_2'].dt.date
print("'date' column (first 5 values):")
print(df_poster['date'].head())
print("-" * 50)

# Step 11: Remove the rows until index 8 (i.e., keep rows from index 8 onward)
# This assumes the DataFrame has at least 8 rows and you want to skip the first 8.
print(f"DataFrame shape before slicing rows: {df_poster.shape}")
if len(df_poster) > 8:
    df_poster = df_poster.iloc[9:].reset_index(drop=True)
    print(f"DataFrame shape after slicing rows (from index 8): {df_poster.shape}")
else:
    print("DataFrame has 8 or fewer rows, skipping row slicing step.")
print(df_poster.head())
print("-" * 50)

# Step 12: Add the 'dap' column (Days After Planting)
# The reference planting date is 5/16/2023.
print("Calculating 'dap' (Days After Planting)...")
planting_date = pd.to_datetime('2023-05-16')
# Ensure 'datetime_pixel_2' is still in datetime format for subtraction
df_poster['dap'] = (df_poster['datetime_pixel_2'] - planting_date).dt.days
print("'dap' column (first 5 values):")
print(df_poster[['date', 'datetime_pixel_2', 'dap']].head())
print("-" * 50)

# Step 13: Display the final transformed DataFrame (optional)
print("Final Transformed DataFrame (first 5 rows):")
print(df_poster.head())

# Optional: Save the transformed DataFrame to a new Excel file
output_file_path = r"D:\Capstone\EDA\transformed_poster_data_with_dap_column.xlsx"
try:
    df_poster.to_excel(output_file_path, index=False)
    print(f"\nTransformed DataFrame saved to {output_file_path}")
except Exception as e:
    print(f"\nError saving to Excel: {e}")

```

```python
import matplotlib.pyplot as plt

# Step 1: Calculate NDVI for each pair of NIR and RED columns
df_grouped['ndvi_pixel_2'] = (df_grouped['pixel_2_nir'] - df_grouped['pixel_2_red']) / (df_grouped['pixel_2_nir'] + df_grouped['pixel_2_red'])
df_grouped['ndvi_red_tarp'] = (df_grouped['red_tarp_nir'] - df_grouped['red_tarp_red']) / (df_grouped['red_tarp_nir'] + df_grouped['red_tarp_red'])
df_grouped['ndvi_white_tarp'] = (df_grouped['white_tarp_nir'] - df_grouped['white_tarp_red']) / (df_grouped['white_tarp_nir'] + df_grouped['white_tarp_red'])

# Step 2: Plot each NDVI column against 'dap'
ndvi_columns = ['ndvi_pixel_2', 'ndvi_red_tarp', 'ndvi_white_tarp']

for column in ndvi_columns:
    plt.figure(figsize=(8, 5))  # Set the figure size
    plt.plot(df_grouped['dap'], df_grouped[column], marker='o', color='g', label=column)
    plt.title(f'{column} vs Days After Planting (DAP)')
    plt.xlabel('Days After Planting (DAP)')
    plt.ylabel('NDVI')
    plt.grid(True)
    plt.ylim([0,1])
    plt.legend()
    plt.show()
```

```python
import statsmodels.api as sm

# Columns to predict (dependent variables)
dependent_columns = ['pixel_2_red', 'pixel_2_nir',
                     'red_tarp_red', 'red_tarp_nir',
                     'white_tarp_red', 'white_tarp_nir',
                     'ndvi_pixel_2', 'ndvi_red_tarp', 'ndvi_white_tarp']

# Dictionary to store regression results
regression_results = {}

# Run regression: each column ~ dap
for col in dependent_columns:
    X = df_grouped[['dap']]  # independent variable
    X = sm.add_constant(X)   # adds intercept term
    y = df_grouped[col]      # dependent variable
    
    model = sm.OLS(y, X).fit()
    regression_results[col] = model
```

```python
# Print regression summaries
for col, model in regression_results.items():
    print(f"Regression summary for '{col}' ~ dap:\n")
    print(model.summary())
    print("\n" + "="*80 + "\n")
```

```python
import matplotlib.pyplot as plt

# Plot regression result for each column
for col in dependent_columns:
    plt.figure(figsize=(8, 5))
    
    # X and predictions
    X = df_grouped[['dap']]
    X_const = sm.add_constant(X)
    y_true = df_grouped[col]
    y_pred = regression_results[col].predict(X_const)

    # Scatter actual data
    plt.scatter(df_grouped['dap'], y_true, color='blue', label='Actual')
    # Regression line
    sorted_idx = df_grouped['dap'].argsort()
    plt.plot(df_grouped['dap'].iloc[sorted_idx], y_pred.iloc[sorted_idx], color='red', label='Regression Line')
    
    plt.title(f'{col} vs DAP (Regression)')
    plt.xlabel('Days After Planting (DAP)')
    plt.ylabel(col)
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
# Plot residuals for each regression
for col in regression_results:
    model = regression_results[col]
    X = df_grouped[['dap']]
    X_const = sm.add_constant(X)
    
    y_true = df_grouped[col]
    y_pred = model.predict(X_const)
    residuals = y_true - y_pred

    plt.figure(figsize=(8, 5))
    plt.scatter(df_grouped['dap'], residuals, color='purple', alpha=0.7)
    plt.axhline(y=0, color='gray', linestyle='--')  # Zero error line
    plt.title(f'Residuals of {col} vs DAP')
    plt.xlabel('Days After Planting (DAP)')
    plt.ylabel('Residuals')
    plt.grid(True)
    plt.show()
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

# Add RMSE to each regression result
regression_summary = []

for dependent_var, model in regression_results.items():
    # Get predicted values
    y_pred = model.fittedvalues
    # Get actual values
    y_actual = df_grouped[dependent_var]
    # Calculate RMSE
    rmse = np.sqrt(mean_squared_error(y_actual, y_pred))

    # Extract other metrics
    intercept = model.params['const']
    slope = model.params['dap']
    r_squared = model.rsquared
    p_value = model.pvalues['dap']

    # Append to summary list
    regression_summary.append({
        'ID': f'{dependent_var} ~ dap',
        'Intercept (β₀)': intercept,
        'Slope (β₁)': slope,
        'R-squared': r_squared,
        'p-value': p_value,
        'RMSE': rmse
    })

# Convert to DataFrame and round
regression_summary_df = pd.DataFrame(regression_summary).round(4)

# Display the table
print(regression_summary_df)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

# Add RMSE to each regression result
regression_summary = []

for dependent_var, model in regression_results.items():
    # Get predicted values
    y_pred = model.fittedvalues
    # Get actual values
    y_actual = df_scaled[dependent_var]
    # Calculate RMSE
    rmse = np.sqrt(mean_squared_error(y_actual, y_pred))

    # Extract other metrics
    intercept = model.params['const']
    slope = model.params['dap']
    r_squared = model.rsquared
    p_value = model.pvalues['dap']

    # Append to summary list
    regression_summary.append({
        'ID': f'{dependent_var} ~ dap',
        'Intercept (β₀)': intercept,
        'Slope (β₁)': slope,
        'R-squared': r_squared,
        'p-value': p_value,
        'RMSE': rmse
    })

# Convert to DataFrame and round
regression_summary_df = pd.DataFrame(regression_summary).round(4)

# Display the table
print(regression_summary_df)
```
