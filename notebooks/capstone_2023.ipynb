```python
# Loading the values of all files using rasterio
def tarp_raster_to_dataframe(raster_files,band_names):
    raster_values = []
    for raster_path in raster_files:
        image_name = os.path.splitext(os.path.basename(raster_path))[0]

        with rasterio.open(raster_path) as src:
            band_data  = src.read()
            raster_values.append({
                'Raster_name': image_name,
                'band_data': band_data
            })



# Define band names in order
# band_names = ['blue', 'green', 'red', 'NIR']
    raster_tarp_records = []
    for item in raster_values:
        Raster_name = item['Raster_name']  # e.g., 'rastername_pixel_1'
        tarp_id = Raster_name.split('_')[-1]  # e.g., '1'
        arr = item['band_data']  # shape: (bands, height, width)
        # Flatten each band and create a dict for this pixel
        row = {'Raster_name': Raster_name}
        for i, band in enumerate(band_names):
            # Flatten the band array and store as a list
            row[f'{tarp_id}_{band}'] = arr[i].flatten().tolist()
        raster_tarp_records.append(row)


    band_keys = [f"{tarp_id}_{band}" for band in band_names]

    # Filter out 65535 values from each band list in red_tarp_records
    # band_keys = ['FID2_blue', 'FID2_green', 'FID2_red', 'FID2_NIR']
    filtered_records = []
    for x in raster_tarp_records:
        for key in band_keys:
            x[key] = x[key][1]
        filtered_records.append(x)
    # Convert to DataFrame
    filtered_records = pd.DataFrame(filtered_records)
    return filtered_records
```

```python
df_red_tarp = tarp_raster_to_dataframe(red_raster_files, band_names)
df_white_tarp = tarp_raster_to_dataframe(white_raster_files, band_names)
```
